#### The "iris" dataset is the well-known universal dataset which is used for the classification problem, in which there are five columns the input features are four: 1. Sepal Length, 2. Sepal Width, 3. Petal Length, 2. Petal Width, and one target variable named: 1. Species.

#### The "iris" dataset is flower dataset which containt the info about a flower length and flower width.

# 01-Importing Essential Libraries:

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
from sklearn.datasets import load_iris
```

# 02-Loading "iris" dataset:

```python
iris = sns.load_dataset("iris")
iris
```

```python
from sklearn.datasets import load_iris

iris1 = load_iris()
```

# 03-Displaying five records of the dataset:

```python
iris.head()
```

# 04-Fetching features variable (X) and target variable (y):

```python
X = iris[["sepal_length", "sepal_width", "petal_length", "petal_width"]]
y = iris["species"]
```

# 05-Training Decision Tree using Gini Criterion:

```python
iris_clf = DecisionTreeClassifier(max_depth=3)
    
iris_clf.fit(X,y)
```

# 06-Visualizing

```python
plt.figure(figsize=(12,7))

plot_tree(
    iris_clf, feature_names=X.columns, class_names=iris_clf.classes_, filled=True, rounded=True)

plt.show()
```

# 07-Prediction on random data:

```python
iris_clf.predict([[1.1, 2.2, 3.3, 4.4]])

#iris_
```
